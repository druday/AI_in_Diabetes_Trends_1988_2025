## Style notes

- **Source of truth:** `scripts/generate_manuscript_figures.py` (this notebook imports it)
- Sans-serif stack (Helvetica Neue / Helvetica / Arial / DejaVu Sans)
- Okabe–Ito–inspired colourblind-safe palette for non-role charts; sequential Blues for heatmaps
- Sankey/alluvial role colors match earlier role-flow assets: `#1f77b4`, `#17becf`, `#ff7f0e`, `#bcbd22`, `#c7c7c7`, with source→target hue blending on transition bands
- Nature-style panel labels **a, b, c** outside top-left; thin spines; muted horizontal grids only
- Double-column (~183 mm) or single-column (~89 mm) widths
- Open orange markers flag the 2016–2019 consortium-inflated network metrics (Fig. 1)
